```python
from pathlib import Path
from urllib.request import urlretrieve

import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

plt.rcParams.update({
    "figure.figsize": (6.5, 3.8), "figure.dpi": 110, "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.alpha": 0.25, "axes.axisbelow": True, "axes.titlelocation": "left", "axes.titleweight": "bold",
})
AZUL, LARANJA, VERDE, CINZA = "#2a78d6", "#eb6834", "#1baf7a", "#6b6b6b"

def titulo_de(ax):
    return ax.get_title(loc="left") or ax.get_title(loc="center") or ax.get_title(loc="right")

URL = (
    "https://raw.githubusercontent.com/americanas-tech/b2w-reviews01/"
    "4639429ec698d7821fc99a0bc665fa213d9fcd5a/B2W-Reviews01.csv"
)
ARQUIVO = Path("B2W-Reviews01.csv")
if not ARQUIVO.exists():
    urlretrieve(URL, ARQUIVO)

dados = pd.read_csv(ARQUIVO, low_memory=False, dtype={"product_id": str}, parse_dates=["submission_date"]).assign(
    nota_5=lambda t: (t["overall_rating"] == 5).astype(int),
    texto=lambda t: t["review_title"].fillna("") + ". " + t["review_text"].fillna(""),
    palavras=lambda t: t["texto"].str.split().str.len(),
    idade=lambda t: (2018 - t["reviewer_birth_year"]).where(lambda s: s.between(14, 100)),
)
TAXA_NOTA_5 = dados["nota_5"].mean()
print(dados.shape, f"· proporção de nota 5: {TAXA_NOTA_5:.1%}")
```

## Questão 4 (2,0 pontos): redesenho

Um colega produziu o gráfico abaixo para um relatório sobre a evolução mensal.

```python
mensal = dados.set_index("submission_date").resample("MS").agg(avaliacoes=("nota_5", "size"), taxa=("nota_5", "mean"))
mensal.index = ["jan", "fev", "mar", "abr", "mai"]

fig, eixo_a = plt.subplots(figsize=(6, 3.4))
eixo_a.bar(mensal.index, mensal["avaliacoes"], color=["red", "orange", "yellow", "green", "blue"])
eixo_b = eixo_a.twinx()                                   # segundo eixo vertical
eixo_b.plot(mensal.index, mensal["taxa"], color="black", marker="o")
eixo_b.set_ylim(0.34, 0.385)
eixo_a.set_title("Gráfico")
plt.show()
```
