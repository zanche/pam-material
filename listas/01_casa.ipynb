Dica: normalize cada linha para norma 1; depois, um único produto de matrizes resolve.

````python
def similaridade_lenta(A):
    n = A.shape[0]
    S = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            S[i, j] = (A[i] * A[j]).sum() / (np.sqrt((A[i] ** 2).sum()) * np.sqrt((A[j] ** 2).sum()))
    return S
````
